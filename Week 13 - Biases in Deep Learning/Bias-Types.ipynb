# Types of Biases

That is a great distinction to make. In data science, the word "bias" is used in two completely different ways, and it’s very easy to mix them up because they both lead to "wrong" answers.

Here is the breakdown of the difference between **Statistical Bias** (the tradeoff) and **Data Bias** (the input).

---

### 1. Statistical Bias (The "Model Simplification" Bias)

This is what the **Bias-Variance Tradeoff** refers to. It is a mathematical property of the algorithm's complexity.

* **The Source:** The model’s architecture.
* **The Cause:** You chose a model that is too simple to capture the true patterns (e.g., using a straight line to map a zig-zagging trend).
* **The Result:** The model "ignores" the data because it isn't flexible enough to follow it.
* **Example:** Predicting house prices using only "square footage" while ignoring location and age. The model is "biased" toward a simple 1-variable explanation.

### 2. Data Bias (The "Garbage In, Garbage Out" Bias)

This refers to **Social or Sampling Bias**. This isn't about the model being too simple; it's about the data being unfair, incomplete, or unrepresentative.

* **The Source:** The training dataset.
* **The Cause:** The data reflects human prejudice, historical inequalities, or a flawed collection process (e.g., only surveying people in one neighborhood).
* **The Result:** The model learns the wrong patterns perfectly. It isn't "underfitting"—it’s accurately reflecting a skewed reality.
* **Example:** A hiring AI that favors male candidates because it was trained on 20 years of historical data where men were hired more often.

---

### Key Comparison

| Feature | Statistical Bias (Tradeoff) | Data Bias (Skewed Input) |
| --- | --- | --- |
| **Is it a math problem?** | Yes, it's about model complexity. | No, it's about data quality/ethics. |
| **How to fix it?** | Use a more complex model (e.g., a Deep Neural Net). | Collect more diverse data or "de-bias" the labels. |
| **The Goal:** | To find the "sweet spot" of accuracy. | To ensure fairness and real-world truth. |
| **Visual Analogy:** | An archer with a stiff bow who can't reach the target. | An archer with a perfect bow, but they're aiming at the wrong target. |

### Why it's confusing

The confusion happens because **Data Bias** can lead to a **Statistical Bias**. If your data only contains one type of person, your model develops a "statistical bias" against other types of people because it "assumes" they don't exist or don't matter based on its training.
